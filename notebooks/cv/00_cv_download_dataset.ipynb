```python
from pathlib import Path

LOCAL_DIR = Path("/content/nexar_collision_prediction")

DRIVE_DIR = Path(
    "/content/drive/MyDrive/AI_Project2/data/cv/nexar_collision_prediction"
)

print("Temporary download path:", LOCAL_DIR)
print("Final Drive path:", DRIVE_DIR)
```

```python
from getpass import getpass
from huggingface_hub import login, HfApi

hf_token = getpass("Paste your HF token: ")

login(
    token=hf_token,
    add_to_git_credential=False
)

username = HfApi().whoami()["name"]
print("Logged in as:", username)

del hf_token
```

```python
import shutil

if LOCAL_DIR.exists():
    shutil.rmtree(LOCAL_DIR)
    print("Old temporary download deleted.")

LOCAL_DIR.mkdir(parents=True, exist_ok=True)

print("Fresh temporary folder:", LOCAL_DIR)
```

```python
from huggingface_hub import snapshot_download

downloaded_path = snapshot_download(
    repo_id="nexar-ai/nexar_collision_prediction",
    repo_type="dataset",
    local_dir=str(LOCAL_DIR),
    token=True,
    force_download=False,
    max_workers=2,
)

print("Download completed:")
print(downloaded_path)
```

```python
for folder_name in ["train", "test-public", "test-private"]:
    folder = LOCAL_DIR / folder_name
    count = len(list(folder.rglob("*.mp4")))

    print(
        folder_name,
        "| exists:", folder.exists(),
        "| videos:", count
    )
```
